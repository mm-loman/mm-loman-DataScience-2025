## Q9. Timing
Compare the runtime of squaring numbers using a Python loop vs. NumPy. Use `%timeit` for each.

```python
nums = np.arange(1000)
%timeit [num**2 for num in nums]
```

```python
%%timeit
result = []
for x in nums:
    result.append(x**2)
```
