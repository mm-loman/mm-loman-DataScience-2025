# 📘 02 - Control Flow

Learn how to use conditional logic in Python.

## ✅ Learning Goals
- Write `if`, `elif`, `else` statements
- Understand truthy/falsy values
- Combine conditions using `and`, `or`, `not`

```python
# 🔧 Basic example
age = 20
if age >= 18:
    print('You are an adult.')
else:
    print('You are a minor.')
```

## 🔀 Multiple Branches

```python
temp = 72
if temp > 90:
    print('Too hot!')
elif temp < 60:
    print('Too cold!')
else:
    print('Just right.')
```
